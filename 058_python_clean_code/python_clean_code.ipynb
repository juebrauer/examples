# Was meint Clean Code?

= guter Code bzw. Code/Sofware von "hoher Qualität"

Was ist aber Code von "hoher Qualität"?

- **Lesbarkeit**: die meisten Software-Entwickler verbringen mehr Zeit damit, Code zu lesen als zu schreiben. Lesbarer Code ist besser verstehbar, benötigt weniger Dokumentation, ist leichter änderbar und erweiterbar

- **Änderbarkeit**: der Code sollte leicht durch mich selber oder andere modifizierbar sein (z.B. durch Austauschen von Implementierungen)

- **Erweiterbarkeit**: der Code sollte leicht durch mich selber oder andere erweiterbar sein (z.B. um neue Funktionalität hinzufügen zu können)

- **Funktionalität**: der Code sollte das machen, was wir uns von ihm wünschen. Daher sind Unittests und Behavior-Tests wichtig.

- **Effizienz**: Code sollte mit genutzten Ressourcen wie Arbeisspeicher oder Rechenzeit sparsam umgehen

- **Prüfbarkeit**: der Code sollte so vorbereitet sein, dass er leicht testbar ist. Zum Beispiel ist es leichter kleinere Units (Funktionen, Klassen) zu testen, die eine Hauptaufgabe haben, als Units, die viele Aufgaben erfüllen.

- **Übertragbarkeit**: der Code sollte so entwickelt sein, dass er zum Beispiel auf Betriebssystemspezifische oder Compilerspezifische Funktionalitäten verzichtet. Er sollte auch auf andere Projekte oder Betriebssysteme übertragbar sein.

- **Zuverlässigkeit/Robustheit**: die Software sollte robust mit fehlerhaften Eingaben umgehen können bzw. auf verschiedenste Probleme vorbereitet sein (kein Speicherplatz mehr auf der Festplatte; Fehlende Schreibrechte; Datenbankverbindung kann nicht aufgebaut werden, weil der Datenbankserver down ist). Selbst wenn ein Totalabsturz der Software auftritt, könnte zum Beispiel durch einen sog. *WatchDog* die Software wieder neu gestartet werden.

- **Benutzbarkeit**: die Software sollte von ihrer Bedienung her intuitiv und leicht erlernbar sein. Aber auch wenn die Software komplexere Funktionalitäten beherbergt und nicht ganz so leicht erlernbar ist, sollte sie zumindestens langfristig effizient hinsichtlich der Bedienung sein

Python hat kein `const` Schlüsselwort!

Über einen größeren Umweg, kann man aber so etwas wie eine Konstante hinbekommen:

```python
class Constants:
    @property
    def PI(self):
        return 3.14159

CONSTANTS = Constants()

# Zugriff auf die Konstante
print(CONSTANTS.PI)

# Versuch, die Konstante zu ändern
CONSTANTS.PI = 42  # Wirft einen Fehler
```

```python
class Constants:
    def __init__(self):
        self._pi = 3.14159  # Interne Variable für PI

    @property
    def PI(self):
        return self._pi

    @PI.setter
    def PI(self, value):
        if value <= 0:
            raise ValueError("PI muss eine positive Zahl sein.")
        self._pi = value

# Verwendung
constants = Constants()

# Zugriff auf die Konstante
print(constants.PI)  # Ausgabe: 3.14159

# Ändern der Konstante (erlaubt durch den Setter)
constants.PI = 3.14
print(constants.PI)  # Ausgabe: 3.14

# Versuch, einen ungültigen Wert zu setzen
try:
    constants.PI = -1  # Löst einen ValueError aus
except ValueError as e:
    print(e)  # Ausgabe: PI muss eine positive Zahl sein.
```

```python
# Unklarer Code: etwas schwer lesbar, nicht alle Fälle abgefangen
def process_order(order):

    if order["processed"] == True:
        return
        
    total = 0
    for i in range(0, len(order["items"])):
        total += order["items"][i]["price"] * order["items"][i]["quantity"]
            
    if total > order["budget"]:
        order["processed"] = True
        order["status"] = "invalid"

    if total < order["budget"]:
        order["processed"] = True
        order["status"] = "valid"
        order["budget"] = order["budget"] - total
        order["items"] = []
```

Exkurs: Unterschied __repr__() und __str__():

Eine hilfreiche Faustregel
- __str__: Wie soll das Objekt für einen Benutzer aussehen?
- __repr__: Wie soll das Objekt für einen Entwickler oder beim Debuggen aussehen?

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"

    def __str__(self):
        return f"({self.x}, {self.y})"

p = Point(3, 4)
print(repr(p))  # Output: Point(x=3, y=4)
print(str(p))   # Output: (3, 4)
print(p)        # Output: (3, 4)
```

# Quizaufgabe: Lesbarkeit mittels Aufteilung in mehrere Funktionen, Bezeichnernamen, etc.

```python
def f(file, term, context_chars=30):
    with open(file, 'r') as f:
        text = f.read()
    os = []
    p = text.lower().find(term.lower())  # Suche nach erstem Vorkommen (case-insensitive)
    while p != -1:
        # Kontext extrahieren
        s = max(0, p - context_chars)
        e = min(len(text), p + len(term) + context_chars)
        b = text[s:p].strip()
        m = text[p:p + len(term)]
        a = text[p + len(term):e].strip()
        os.append({
            "b": b,
            "m": m,
            "a": a,
        })        
        p = text.lower().find(term.lower(), p + len(term))
    for o in os:
        b = o["b"]
        m = o["m"]
        a = o["a"]
        print(b,m,a)
f("example.txt", "Python")
```

```python
def read_file(file_path):
    """
    Liest den Inhalt einer Datei und gibt ihn als String zurück.
    """
    with open(file_path, 'r') as file:
        return file.read()


def find_term_occurrences(text, term, context_chars=30):
    """
    Findet alle Vorkommen eines Suchbegriffs im Text mit Kontext.

    :param text: Der gesamte Text.
    :param term: Der Suchbegriff (case-insensitive).
    :param context_chars: Anzahl der Zeichen vor und nach dem Begriff im Kontext.
    :return: Liste von Kontextinformationen (vorher, Begriff, nachher).
    """
    occurrences = []
    search_start = 0
    term_lower = term.lower()
    text_lower = text.lower()

    while True:
        position = text_lower.find(term_lower, search_start)
        if position == -1:
            break

        # Kontext extrahieren
        start = max(0, position - context_chars)
        end = min(len(text), position + len(term) + context_chars)
        before = text[start:position].strip()
        match = text[position:position + len(term)]
        after = text[position + len(term):end].strip()

        occurrences.append({"before": before, "match": match, "after": after})

        search_start = position + len(term)

    return occurrences


def print_occurrences(occurrences):
    """
    Gibt alle gefundenen Vorkommen mit Kontext aus.
    """
    for occurrence in occurrences:
        print(occurrence["before"], occurrence["match"], occurrence["after"])


def search_term_in_file(file_path, term, context_chars=30):
    """
    Hauptfunktion zum Suchen und Anzeigen von Begriffsvorkommen in einer Datei.
    """
    text = read_file(file_path)
    occurrences = find_term_occurrences(text, term, context_chars)
    print_occurrences(occurrences)


# Beispielaufruf
search_term_in_file("example.txt", "Python")
```

# SOLID-Prinzipien von Robert C. Martin

## Single Responsibility

Eine Funktion, Klasse, Modul sollte nur eine Verantwortung haben.

Warum wichtig? Wenn sich Anforderungen ändern, ist es einfacher, den Code anzupassen, ohne unerwünschte Seiteneffekte zu erzeugen.


```python
class ReportGenerator:
    def generate_report(self):
        # Generiere den Bericht
        pass

    def send_email(self):  # Verletzung von SRP
        # Sende den Bericht per E-Mail
        pass
```

Das bedeutet nicht, dass eine Klasse nur eine Methode haben darf!

Es bedeutet vielmehr, dass eine Klasse nur einen Verantwortungsbereich übernehmen sollte.

Zum Beispiel wäre folgende Klasse absolut ok.

InvoiceCalculator kümmert sich mit den folgenden 3 Methoden nur um Rechnungsberechnungen.

Alles andere, wie z. B. das Speichern oder Darstellen der Rechnung, würde in anderen Klassen erfolgen

```python
class InvoiceCalculator:
    def calculate_total(self, items):
        total = sum(item['price'] * item['quantity'] for item in items)
        return total

    def calculate_discount(self, total, discount_rate):
        return total * discount_rate / 100

    def calculate_tax(self, total, tax_rate):
        return total * tax_rate / 100

# Artikelliste für die Rechnung
items = [
    {'name': 'Book', 'quantity': 2, 'price': 15.0},
    {'name': 'Pen', 'quantity': 5, 'price': 2.5},
    {'name': 'Notebook', 'quantity': 1, 'price': 7.0},
]

# Rabatt- und Steuersätze
discount_rate = 10  # in Prozent
tax_rate = 19  # in Prozent

# Verwendung der InvoiceCalculator-Klasse
calculator = InvoiceCalculator()

# Berechnungen durchführen
total = calculator.calculate_total(items)
discount = calculator.calculate_discount(total, discount_rate)
total_after_discount = total - discount
tax = calculator.calculate_tax(total_after_discount, tax_rate)
grand_total = total_after_discount + tax

# Ergebnis ausgeben
print(f"Zwischensumme: {total:.2f} €")
print(f"Rabatt ({discount_rate}%): -{discount:.2f} €")
print(f"Nach Rabatt: {total_after_discount:.2f} €")
print(f"Steuer ({tax_rate}%): +{tax:.2f} €")
print(f"Gesamtsumme: {grand_total:.2f} €")
```

## Open/Closed Principle

Software-Entitäten (Klassen, Funktionen, Module) sollten offen für Erweiterung, aber geschlossen für Modifikation sein.

Bedeutung: Der bestehende Code sollte nicht geändert werden müssen, wenn neue Funktionalität hinzukommt.

Warum wichtig? Vermeidet Fehler durch Änderungen am existierenden Code und fördert die Erweiterbarkeit.

```python
class Rectangle:
    def area(self):
        # Berechnung der Fläche
        pass

class Circle:
    def area(self):
        # Berechnung der Fläche
        pass

class AreaCalculator:
    def calculate(self, shapes):
        return sum([shape.area() for shape in shapes])

# Hier könnten jetzt neue Formen (z. B. Triangle) hinzugefügt werden,
# ohne die Klasse AreaCalculator ändern zu müssen.
```

### Negativ-Bsp

In diesem Beispiel haben wir eine PaymentProcessor-Klasse, die direkt verschiedene Zahlungsarten behandelt. Jede neue Zahlungsart erfordert Änderungen an der Klasse:

```python
class PaymentProcessor:

    def __init__(self):
        self.total_amount_payed = 0
         
    def process_payment(self, payment_method, amount):
        if payment_method == "credit_card":
            print(f"Processing credit card payment of {amount}")
            self.total_amount_payed += amount
        elif payment_method == "paypal":
            print(f"Processing PayPal payment of {amount}")
            self.total_amount_payed += amount
        else:
            raise ValueError("Unsupported payment method")

# Nutzung
processor = PaymentProcessor()
processor.process_payment("credit_card", 100)
processor.process_payment("paypal", 50)
processor.total_amount_payed
```

### Positiv-Bsp

```python
from abc import ABC, abstractmethod

# Abstrakte Strategie
class PaymentMethod(ABC):
    @abstractmethod
    def process(self, amount):
        pass

# Konkrete Strategien
class CreditCardPayment(PaymentMethod):
    
    def process(self, amount):
        print(f"Processing credit card payment of {amount}")
        return amount

class PayPalPayment(PaymentMethod):
    
    def process(self, amount):
        print(f"Processing PayPal payment of {amount}")
        return amount

# Kontext-Klasse
class PaymentProcessor:
    
    def __init__(self):
        self.total_amount_payed = 0

    def process_payment(self, paymethod : PaymentMethod, amount : float):
        self.total_amount_payed += paymethod.process(amount)

# Nutzung
processor = PaymentProcessor()
processor.process_payment(PayPalPayment(), 100)
processor.process_payment(CreditCardPayment(), 900)
processor.total_amount_payed
```

## Liskov Substitution Principle (LSP)

Ein Objekt einer Unterklasse sollte ohne Probleme anstelle seiner Basisklasse verwendet werden können.

Bedeutung: Unterklassen sollten alle Eigenschaften und Methoden der Basisklasse sinnvoll übernehmen und gegebenenfalls erweitern, ohne deren Verhalten zu ändern.
Warum wichtig? Verhindert Überraschungen im Verhalten von Objekten und erleichtert die Wiederverwendung von Code.

### Positiv-Beispiel

```python
import pandas as pd

class DataFrameProcessor:
    def process(self, df: pd.DataFrame) -> pd.DataFrame:
        # Standardverarbeitung: Entferne NaN-Werte
        return df.dropna()

class SumDataFrameProcessor(DataFrameProcessor):
    def process(self, df: pd.DataFrame) -> pd.DataFrame:
        # Verarbeite und berechne die kumulierte Summe
        processed_df = super().process(df).copy()  # Sicherstellen, dass eine Kopie bearbeitet wird
        processed_df.loc[:, 'cumulative_sum'] = processed_df['values'].cumsum()
        return processed_df

# Nutzung
def run_processor(processor: DataFrameProcessor, df: pd.DataFrame):
    processed_df = processor.process(df)
    print(processed_df)

data = {'values': [1, 2, None, 4]}
df = pd.DataFrame(data)

# Standardverarbeitung
processor1 = DataFrameProcessor()
run_processor(processor1, df)

# Verarbeitung durch SumDataFrameProcessor
processor2 = SumDataFrameProcessor()
run_processor(processor2, df)
```

## Interface Segregation Principle (ISP) - Schnittstellentrennungsprinzip

Eine Klasse sollte keine Schnittstellen implementieren müssen, die sie nicht benötigt.

Bedeutung: Große, allgemeine Interfaces (Schnittstellen) sollten in kleinere, spezifischere Interfaces zerlegt werden.
Warum wichtig? Verhindert, dass Klassen Methoden implementieren müssen, die für sie irrelevant sind.

```python
class Animal:
    def fly(self):  # Nicht jedes Tier kann fliegen
        pass

    def swim(self):
        pass
```

```python
# Stattdessen sollten separate Schnittstellen wie Flyable
# und Swimmable verwendet werden:

class Flyable:
    def fly(self):
        pass

class Swimmable:
    def swim(self):
        pass

class Dog(Swimmable):
    def swim(self):
        print("Ja! Auch Hunde können schwimmen!")

class Duck(Flyable, Swimmable):

    def swim(self):
        print("Klaro! Enten schwimmen gerne!")

    def fly(self):
        print("Und na ja... Enten können auch ein wenig fliegen.")
```

```python
do = Dog()
do.swim()
```

```python
du = Duck()
du.swim()
du.fly()
```

## Dependency-Inversion-Principle

heißt: Eine allgemeine Klasse hängt von einer anderen speziellen Klasse ab.

Es sollte doch anders herum sein!

High-Level-Module sollten nicht von Low-Level-Modulen abhängen. Beide sollten von abstrakten Schnittstellen abhängen.

Bedeutung: Abhängigkeiten sollten auf Abstraktionen basieren, nicht auf konkreten Implementierungen.

Warum wichtig? Fördert flexible und testbare Architekturen

```python
# Beispiel: Verletzung

# Das ist der Kern des Dependency Inversion Principle (DIP):
# High-Level-Klassen wie Switch sollten nicht von Low-Level-Klassen
# wie LightBulb abhängen, sondern beide sollten von einer abstrakten
# Schnittstelle (LightSource) abhängen.

class LightBulb:
    def turn_on(self):
        pass

    def turn_off(self):
        pass

class Switch:
    def __init__(self, bulb: LightBulb):  # Abhängigkeit von einer konkreten Klasse
        self.bulb = bulb

    def operate(self, state):
        if state == "ON":
            self.bulb.turn_on()
        else:
            self.bulb.turn_off()
```

```python
# Lösung

from abc import ABC, abstractmethod

class LightSource(ABC):  # Abstrakte Klasse
    @abstractmethod
    def turn_on(self):
        pass

    @abstractmethod
    def turn_off(self):
        pass

class LightBulb(LightSource):  # Konkrete Implementierung
    def turn_on(self):
        print("LightBulb is on")

    def turn_off(self):
        print("LightBulb is off")

class LEDLight(LightSource):  # Konkrete Implementierung
    def turn_on(self):
        print("LEDLight is on")

    def turn_off(self):
        print("LEDLight is off")

class Switch:
    def __init__(self, source: LightSource):  # Abhängigkeit von der Abstraktion
        self.source = source

    def operate(self, state):
        if state == "ON":
            self.source.turn_on()
        else:
            self.source.turn_off()
```

### Negativ-Bsp

Eine Anwendung benötigt historische Aktienkursdaten. Die Geschäftslogik soll unabhängig von der spezifischen Implementierung des Datenabfragedienstes sein, sodass wir den Datenanbieter leicht austauschen können.

```python
import yfinance as yf

class StockAnalyzer:
    def get_historical_data(self, ticker: str, start_date: str, end_date: str):
        stock = yf.Ticker(ticker)
        data = stock.history(start=start_date, end=end_date)
        return data

# Nutzung
analyzer = StockAnalyzer()
data = analyzer.get_historical_data("AAPL", "2023-01-01", "2023-12-31")
print(data)
```

### Positiv-Bsp

```python
from abc import ABC, abstractmethod
import yfinance as yf
import pandas as pd

# Abstrakte Schnittstelle
class StockDataService(ABC):
    @abstractmethod
    def get_historical_data(self, ticker: str, start_date: str, end_date: str) -> pd.DataFrame:
        pass

# Implementierung mit yfinance
class YahooFinanceService(StockDataService):
    def get_historical_data(self, ticker: str, start_date: str, end_date: str) -> pd.DataFrame:
        stock = yf.Ticker(ticker)
        return stock.history(start=start_date, end=end_date)

# Geschäftslogik
class StockAnalyzer:
    def __init__(self, data_service: StockDataService):
        self.data_service = data_service

    def analyze(self, ticker: str, start_date: str, end_date: str):
        data = self.data_service.get_historical_data(ticker, start_date, end_date)
        # Beispiel: Berechnung des durchschnittlichen Schlusskurses
        avg_close = data['Close'].mean()
        print(f"Average closing price for {ticker}: {avg_close:.2f}")

# Nutzung
yahoo_service = YahooFinanceService()
analyzer = StockAnalyzer(yahoo_service)
analyzer.analyze("AAPL", "2023-01-01", "2023-12-31")
```
