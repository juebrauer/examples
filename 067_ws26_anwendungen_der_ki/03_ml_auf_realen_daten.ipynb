```python
import kagglehub
import pandas

# 1. Datensatz runterladen und einlesen
path = kagglehub.dataset_download("wspirat/germany-used-cars-dataset-2023")
df_raw = pandas.read_csv(f"{path}/data.csv")

# 2. Bereinigung der Daten
df_clean = df_raw[ ["mileage_in_km", "power_ps", "price_in_euro"] ].copy()
df_clean["price_in_euro"] = pandas.to_numeric(df_clean["price_in_euro"],
                                              errors="coerce")
df_clean
```
