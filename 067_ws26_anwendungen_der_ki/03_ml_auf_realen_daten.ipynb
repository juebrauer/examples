# Alles in einer Zelle mit ML

```python
import pandas
import kagglehub

from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.tree import export_text
from sklearn.metrics import mean_absolute_percentage_error


# 1. Datensatz runterladen und einlesen
path = kagglehub.dataset_download("wspirat/germany-used-cars-dataset-2023")
df_raw = pandas.read_csv(f"{path}/data.csv")

# 2. Bereinigung der Daten
spalten = ["mileage_in_km", "power_ps", "price_in_euro", "brand"]
df_clean = df_raw[ spalten ].copy()
for spalte in spalten:
    df_clean[spalte] = pandas.to_numeric(df_clean[spalte],
                                         errors="coerce")
# 3. Unvollständige Zeilen in der Tabelle verwerfen
df_clean = df_clean.dropna()

# 4. Daten fürs ML vorbereiten

# 4.1 Was ist Input x? Was ist Output y?
x = df_clean[ ["mileage_in_km", "power_ps"] ]
y = df_clean[ ["price_in_euro"] ]

# 4.2 Train-Test-Split
x_train, x_test, y_train, y_test = train_test_split(x,y,
                                          test_size=0.2)

# 5. Einen DecisionTree (DT) bauen
model = DecisionTreeRegressor()
model = model.fit(x_train, y_train)

# 6. Das Modell visualisieyren
txt = export_text(model, feature_names=x_train.columns)
print(txt[:2000])

# 7. Das Modell testen
y_pred = model.predict( x_test )
y_true = y_test["price_in_euro"].values
MAPE = mean_absolute_percentage_error(y_true, y_pred) * 100
print(f"MAPE={MAPE:.2f}%")
```

```python
THRESHOLD = 2000

correct = 0
for i in range(len(y_pred)):
    a = y_pred[i]
    b = y_true[i]
    c = abs(a-b)
    d = c < THRESHOLD
    if d:
        correct += 1
    #print(f"{a:10.1f} {b:10.1f} {c:10.1f} {d}")
```
