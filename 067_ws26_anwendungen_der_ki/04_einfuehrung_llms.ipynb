```python
from openai import OpenAI

my_key = "hier eintragen"


anfrage1 = """Ich wohne jetzt woanders. Wo muss ich das bei Ihnen eingeben?
Wohne jetzt in der Nebenstraße, nicht mehr in der Hauptstraße.
"""

anfrage2 = """Das doofe Ding läuft ständig aus. Die Wäsche wird auch nicht trocken!
Es ist doch gerade mal 3 Monate her, dass ich das bei euch gekauft habe.
Ich will eine neue!"""

prompt = f"""
         Du bist ein Assistent für die Klassifikation von Kundenanfragen.
         Jede Kundenanfrage soll in einer von den folgenden 4 Klassen
         einsortiert werden:
         1.) Garantieanspruch
         2.) Adressänderung
         3.) Produktinformationsanfrage
         4.) Alles andere

         Hier die Anfrages eines Kunden:
         {anfrage2}

         Antworte nur mit einer Zahl 1-4
         """


client = OpenAI(api_key=my_key)

response = client.responses.create(model="gpt-6-luna",
                                   input=prompt)

print("Unser Gesamtprompt:")
print(prompt)

print("-" * 50)

print(response.output_text)
```
