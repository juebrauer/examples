```python
import matplotlib.pyplot as plt

obst = ["Äpfel", "Bananen", "Orangen", "Trauben"]
anzahl = [8, 5, 4, 3]

plt.pie(anzahl, labels=obst, autopct="%1.1f%%", startangle=90)
plt.title("Lieblingsobst")
plt.axis("equal")  # sorgt für einen runden Kreis
plt.show()
```
