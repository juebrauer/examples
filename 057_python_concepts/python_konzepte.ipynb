# Decorators

Ein Decorator ist eine Funktion, die eine andere Funktion erweitert oder verändert, ohne deren Quellcode direkt zu ändern.

Typische Anwendungen für Decorators:
- Protokollierung
- Zeitmessung
- Wiederholungslogik

Die Schreibweise:

        @mein_decorator
        def meine_funktion():
            ...

ist quasi Kurzschreibweise für:

        meine_funktion = mein_decorator(meine_funktion)


````python
def f():
    print("Hallo")

# Funktionen können wie andere Objekte
# an Funktionen als Parameter übergeben werden!
def g(x):
    print("Vor dem Aufruf der übergebenen Funktion")
    x()
    print("Nach dem Aufruf der übergebenen Funktion")

g(f)
````

````python
type(f), type(g)
````

````python
def g(x):

    def wrapper():
        print("Vor dem Aufruf der übergebenen Funktion")
        x()
        print("Nach dem Aufruf der übergebenen Funktion")

    return wrapper

@g
def f():
    print("Hallo")


f()
````

````python
type(f), type(g)
````

## Anwendungsbeispiel 1: Zeitmessung

````python
import time


def measure_time(x):

    def wrapper():
        timestamp1 = time.time()        
        x()
        timestamp2 = time.time()
        # Verstrichene Wanduhrzeit vs. CPU-Zeit
        # time.time(): Wanduhrzeit
        # time.process_time(): CPU-Zeit
        duration = timestamp2 - timestamp1
        print(f"Execution time for function x(): {duration} seconds")

    return wrapper


@measure_time
def f():
    zahlen = []
    for i in range(50_000_000):
        zahlen.append(i**2)    

f()
````

## Zu dekorierende Funktion mit Parametern

````python
import time


def measure_time(x):

    def wrapper(*args, **kwargs):
        timestamp1 = time.time()

        result = x(*args, **kwargs)

        timestamp2 = time.time()
        duration = timestamp2 - timestamp1
        print(f"Execution time for function x(): {duration} seconds")
        return result

    return wrapper


@measure_time
def f(start, stop):
    zahlen = []
    for i in range(start, stop+1):
        zahlen.append(i**2)
    return zahlen[-1]

f(1_000_000, 9_500_000)
````

## Dekoratoren mit Parametern

````python
def repeat(times, text):

    def decorator(func):        

        def wrapper(*args, **kwargs):

            print(f"{text} {times}x")
            result = None
            for run in range(times):
                print(f"Run: {run}")
                result = func(*args, **kwargs)

            return result

        return wrapper

    return decorator


def measure_time(x):

    def wrapper(*args, **kwargs):
        timestamp1 = time.time()

        result = x(*args, **kwargs)

        timestamp2 = time.time()
        duration = timestamp2 - timestamp1
        print(f"Execution time for function x(): {duration} seconds")
        return result

    return wrapper



@repeat(3, "Repeating")
@measure_time
def f(start, stop):
    zahlen = []
    for i in range(start, stop+1):
        zahlen.append(i**2)
    return zahlen[-1]

f(1_000_000, 10_000_000)
````

## Der Dekorator @wraps

````python
import time
from functools import wraps


def measure_time(x):

    @wraps(x)
    def wrapper(*args, **kwargs):
        timestamp1 = time.time()

        result = x(*args, **kwargs)

        timestamp2 = time.time()
        duration = timestamp2 - timestamp1
        print(f"Execution time for function x(): {duration} seconds")
        return result

    return wrapper


@measure_time
def f(start, stop):
    """
    This function generates squares for some numbers.
    """
    zahlen = []
    for i in range(start, stop+1):
        zahlen.append(i**2)
    return zahlen[-1]

f(1_000_000, 9_500_000)

print(f.__name__)
print(f.__doc__)
````

## Anwendungsbeispiel 2: Logging

````python
from functools import wraps
from datetime import datetime


def log_call(func):

    @wraps(func)
    def wrapper(*args, **kwargs):
        zeitpunkt = datetime.now().strftime("%H:%M:%S")

        print(f"[{zeitpunkt}] Funktion {func.__name__} wird aufgerufen")
        print(f"  Positionsargumente: {args}")
        print(f"  Keywordargumente:   {kwargs}")

        result = func(*args, **kwargs)

        print(f"  Rückgabewert:        {result}\n")

        return result

    return wrapper


@log_call
def addiere(a, b):
    return a + b


@log_call
def begruessung(name, formell=False):
    if formell:
        return f"Guten Tag, {name}."
    return f"Hallo, {name}!"


addiere(4, 7)
begruessung("Jürgen", formell=True)
````

## Dekoratoren bei Klassen (ohne Parameter)

````python
def add_description(cls):
    def describe(self):
        return f"Ich bin ein Objekt der Klasse {cls.__name__}."

    cls.describe = describe
    return cls


@add_description
class Person:
    def __init__(self, name):
        self.name = name


person = Person("Anna")

print(person.name)
print(person.describe())
````

## Dekorator bei Klassen (mit Parameter)

````python
def add_category(category):
    def decorator(cls):
        cls.category = category
        return cls

    return decorator


@add_category("Fahrzeug")
class Car:
    def __init__(self, brand):
        self.brand = brand


car = Car("Volvo")

print(car.brand)
print(car.category)
````

# Klassen-, Instanz- und statische Methoden

In Python gibt es drei typische Arten von Methoden:

- Instanzmethoden
- Klassenmethoden
- statische Methoden

Der wichtigste Unterschied ist, worauf die Methode automatisch Zugriff erhält.

````python
class User:
    anzahl_user = 0

    def __init__(self, name, alter):
        self.name = name
        self.alter = alter
        User.anzahl_user += 1


    # Zugriff auf Daten eines Objektes --> wir nehmen eine Instanzmethode
    def vorstellen(self):
        print(f"Ich bin {self.name} und {self.alter} Jahre alt.")

    # Zugriff auf Daten der Klassen --> wir nehmen eine Klassenmethode
    @classmethod
    def zeige_anzahl(cls):
        print(f"Anzahl der User: {cls.anzahl_user}")

    @classmethod
    def aus_text(cls, text):
        name, alter = text.split(",")
        return cls(name, int(alter))
    
    @staticmethod
    def aus_text2(text):
        name, alter = text.split(",")
        return User(name, int(alter))

    # Die Funktion braucht weder Daten eines Objektes noch Daten der Klasse --> statische Methode
    @staticmethod
    def ist_volljaehrig(alter):
        return alter >= 18


user1 = User("Anna", 24)
user2 = User("Ben", 16)

user1.vorstellen()
user2.vorstellen()

User.zeige_anzahl()

print(User.ist_volljaehrig(24))
print(User.ist_volljaehrig(16))

user = User.aus_text2("Jürgen,50")
print(user.name)
print(user.alter)
````

## Unterschied: Klassen- vs. statische Methode

````python
class User:
    def __init__(self, name):
        self.name = name

    @classmethod
    def aus_text(cls, text):
        # cls ist die Klasse, über die die Methode aufgerufen wurde
        return cls(text)

    @staticmethod
    def aus_text_statisch(text):
        # User ist hier fest einprogrammiert
        return User(text)


class Admin(User):
    pass


objekt1 = Admin.aus_text("Anna")
objekt2 = Admin.aus_text_statisch("Ben")

print(objekt1.name)
print(type(objekt1))  # Admin

print(objekt2.name)
print(type(objekt2))  # User
````

## Merkregeln: Wann welche Art von Methode?

| Methodenart           | Dekorator       | Erster Parameter | Automatischer Zugriff                              | Typische Verwendung                                                                                | Aufruf                            |
| --------------------- | --------------- | ---------------- | -------------------------------------------------- | -------------------------------------------------------------------------------------------------- | --------------------------------- |
| **Instanzmethode**    | keiner          | `self`           | Auf die konkrete Instanz und ihre Instanzvariablen | Wenn die Methode mit den Daten eines bestimmten Objekts arbeitet                                   | `user.vorstellen()`               |
| **Klassenmethode**    | `@classmethod`  | `cls`            | Auf die Klasse und ihre Klassenvariablen           | Wenn die Methode mit gemeinsamen Klassendaten arbeitet oder alternative Konstruktoren bereitstellt | `User.zeige_anzahl()`             |
| **Statische Methode** | `@staticmethod` | keiner           | Weder auf Instanz noch Klasse                      | Wenn eine Hilfsfunktion logisch zur Klasse gehört, aber keine Objekt- oder Klassendaten benötigt   | `User.ist_gueltiger_name("Anna")` |


| Frage                                                                                 | Geeignete Methodenart                                 |
| ------------------------------------------------------------------------------------- | ----------------------------------------------------- |
| Benötigt die Methode Daten eines konkreten Objekts, zum Beispiel `name` oder `alter`? | **Instanzmethode**                                    |
| Benötigt die Methode Zugriff auf Klassenvariablen, zum Beispiel `anzahl_user`?        | **Klassenmethode**                                    |
| Soll die Methode eine neue Instanz auf besondere Weise erzeugen?                      | **Klassenmethode**                                    |
| Benötigt die Methode weder `self` noch `cls`, gehört aber thematisch zur Klasse?      | **Statische Methode**                                 |
| Ist die Funktion auch ohne die Klasse sinnvoll und gehört nicht wirklich zu ihr?      | Besser eine **normale Funktion außerhalb der Klasse** |


| Schreibweise                          | Bedeutung                           |
| ------------------------------------- | ----------------------------------- |
| `def methode(self):`                  | Arbeitet mit einem konkreten Objekt |
| `@classmethod`<br>`def methode(cls):` | Arbeitet mit der Klasse             |
| `@staticmethod`<br>`def methode():`   | Benötigt weder Objekt noch Klasse   |

````python
import os

class DateiIterator:
    def __init__(self, verzeichnis, dateityp):
        self.verzeichnis = verzeichnis
        self.dateityp = dateityp
        self.dateiliste = []
        self.index = -1
        for root, dirs, files in os.walk(verzeichnis):
            for datei in files:
                if datei.endswith('.' + dateityp):
                    self.dateiliste.append(os.path.join(root, datei))

    def __iter__(self):
        return self

    def __next__(self):
        self.index += 1
        if self.index >= len(self.dateiliste):
            raise StopIteration
        return self.dateiliste[self.index]

iterator = DateiIterator("project", "csv")
for datei in iterator:
    print(datei)
````

````python
import numpy as np
import pandas as pd
from pathlib import Path

ordner = Path("messwerte")
ordner.mkdir(exist_ok=True)

spalten = ["Druck", "Temperatur", "Drehmoment1", "Drehmoment2", "Durchfluss"]

for i, anzahl in enumerate([100, 150, 80, 200, 120], start=1):
    df = pd.DataFrame(
        np.random.rand(anzahl, len(spalten)),
        columns=spalten,
        index=pd.date_range(
            start=f"2026-07-{i:02d} 08:00:00",
            periods=anzahl,
            freq="1s"
        )
    )

    df.index.name = "Zeitstempel"
    df.to_csv(ordner / f"messung_{i}.csv")
````

````python
import pandas as pd


class ColumnMaster:
    def __init__(self, spaltenname):
        self.spaltenname = spaltenname
        self.messwerte = pd.Series(dtype=float)

    def __add__(self, dateiname):
        neue_werte = pd.read_csv(dateiname)[self.spaltenname]
        self.messwerte = pd.concat(
            [self.messwerte, neue_werte],
            ignore_index=True
        )
        return self

    def __len__(self):
        return len(self.messwerte)

    def __iter__(self):
        return iter(self.messwerte)

    def __getitem__(self, index):
        return self.messwerte.iloc[index]

    def __str__(self):
        return f"{len(self)} Messwerte der Spalte '{self.spaltenname}'"
````

````python
from pathlib import Path

temperaturen = ColumnMaster("Temperatur")

for datei in Path("messwerte").glob("*.csv"):
    temperaturen + datei

print(temperaturen)
print(f"Anzahl der Messwerte: {len(temperaturen)}")
print(f"Erster Messwert: {temperaturen[0]:.2f}")

print(f"Mittelwert: {temperaturen.messwerte.mean():.2f}")
print(f"Minimum: {temperaturen.messwerte.min():.2f}")
print(f"Maximum: {temperaturen.messwerte.max():.2f}")
````

````python
customer = {
    "customer_id": 4711,
    "name": "Anna Müller",
    "status": "new"
}

match customer["status"]:
    case "new":
        print("Willkommens-E-Mail senden.")

    case "active" | "premium":
        print("Der Kunde darf Bestellungen aufgeben.")

    case "blocked" | "suspended":
        print("Das Kundenkonto darf momentan nicht verwendet werden.")

    case "deleted":
        print("Das Kundenkonto wurde gelöscht.")

    case "new":
        print("Hatten wir das nicht schon?")

    case _:
        print("Unbekannter Kundenstatus.")
````

## Klassische Getter- und Setter-Methoden

````python
class Person:
    def __init__(self, name, alter):
        self._name = name
        self._alter = alter

    def get_name(self):
        return self._name

    def set_name(self, name):
        self._name = name

    def get_alter(self):
        return self._alter

    def set_alter(self, alter):
        if alter < 0:
            raise ValueError("Das Alter darf nicht negativ sein.")

        self._alter = alter



person = Person("Anna", 25)

print(person.get_name())

person.set_alter(26)
print(person.get_alter())
````

## Die Python-übliche Lösung: property

Python bietet Properties. Dadurch sieht der Zugriff von außen wie ein normaler Attributzugriff aus, intern werden aber Getter- und Setter-Methoden ausgeführt.

````python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

    @property
    def alter(self):
        return self._alter

    @alter.setter
    def alter(self, neuer_wert):
        if neuer_wert < 0:
            raise ValueError("Das Alter darf nicht negativ sein.")

        self._alter = neuer_wert

person = Person("Anna", 25)
print(person.alter)  # ruft den Getter auf
person.alter = 26    # ruft den Setter auf
````

````python
class Kreis:
    def __init__(self, radius):
        self.radius = radius

    @property
    def flaeche(self):
        return 3.14159 * self.radius ** 2

kreis = Kreis(5)
print(kreis.flaeche)

# flaeche sieht wie ein Attribut aus, wird aber bei jedem Zugriff berechnet.
````

## Wann sollte man Properties verwenden?

Properties sind sinnvoll, wenn beim Zugriff zusätzliche Logik nötig ist, zum Beispiel:

- Werte validieren
- Werte umrechnen
- Werte berechnen
- Änderungen protokollieren
- veraltete Schnittstellen kompatibel halten
- Attribute schreibgeschützt machen

## Vorteil von properties: Code muss nicht geändert werden

````python
class Temperatur:
    def __init__(self, celsius):
        self.celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, wert):
        if wert < -273.15:
            raise ValueError(
                "Die Temperatur liegt unter dem absoluten Nullpunkt."
            )

        self._celsius = wert
````

Der bestehende Benutzercode muss nicht geändert werden:

        temperatur.celsius = 20

Wir müssen hier also z.B. nicht neuerdings eine setter-Funktion mit () aufrufen, wenn wir eine Bereichsüberprüfung einbauen wollen.

Das ist einer der Hauptgründe, warum Python direkten Attributzugriff bevorzugt: Bei Bedarf kann später eine Property ergänzt werden.

## Gleichzeitig ins Terminal und in eine Datei loggen

Dafür verwendet man sogenannte Handler.

Ein Logger erzeugt eine Meldung. Die Handler entscheiden, wohin sie geschrieben wird.

        Logger
        ├── StreamHandler  → Terminal
        └── FileHandler    → Datei

````python
import logging

logger = logging.getLogger("logger1")
logger.setLevel(logging.DEBUG)

formatter = logging.Formatter(
    "%(asctime)s | %(levelname)-8s | %(name)s | %(message)s"
)

console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)
console_handler.setFormatter(formatter)

file_handler = logging.FileHandler(
    "application.log",
    encoding="utf-8",
)
file_handler.setLevel(logging.DEBUG)
file_handler.setFormatter(formatter)

logger.addHandler(console_handler)
logger.addHandler(file_handler)

logger.debug("Details für die Logdatei")
logger.info("Programm gestartet")
logger.warning("Eine Warnung")
````

## Wachsende Logdateien begrenzen

Ein normaler FileHandler schreibt unbegrenzt in dieselbe Datei.
Mit einem RotatingFileHandler kann man die Größe begrenzen.

````python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger(__name__)
logger.setLevel(logging.DEBUG)

handler = RotatingFileHandler(
    "application.log",
    maxBytes=100,
    backupCount=3,
    encoding="utf-8",
)

formatter = logging.Formatter(
    "%(asctime)s | %(levelname)s | %(message)s"
)
handler.setFormatter(formatter)

logger.addHandler(handler)

for i in range(100):
    logger.info("Programm gestartet")
````

## Konfiguration für ein kleines Projekt

Eine Funktion kann die Logging-Konfiguration zentral einrichten:

````python
import logging
from logging.handlers import RotatingFileHandler


def configure_logging(logger_name):
    formatter = logging.Formatter(
        "%(asctime)s | %(levelname)-8s | "
        "%(name)s | %(message)s"
    )

    console_handler = logging.StreamHandler()
    console_handler.setLevel(logging.INFO)
    console_handler.setFormatter(formatter)

    file_handler = RotatingFileHandler(
        "application.log",
        maxBytes=1_000_000,
        backupCount=3,
        encoding="utf-8",
    )
    file_handler.setLevel(logging.DEBUG)
    file_handler.setFormatter(formatter)

    logger = logging.getLogger(logger_name)
    logger.setLevel(logging.DEBUG)
    logger.addHandler(console_handler)
    logger.addHandler(file_handler)
````

Verwendung:

````python
configure_logging("logger1") # nur einmal aufrufen
````

## Logfile Viewer

z.B. klogg

````python
import logging
import random

logging.basicConfig(
    filename="application.log",
    filemode="w",
    level=logging.DEBUG,
    format="%(asctime)s | %(levelname)-8s | %(message)s"
)

messages = [
    (logging.DEBUG, "Lade Konfiguration"),
    (logging.INFO, "Benutzer angemeldet"),
    (logging.WARNING, "Speicher wird knapp"),
    (logging.ERROR, "Verbindung fehlgeschlagen"),
    (logging.CRITICAL, "Systemfehler"),
]

for i in range(20_000):
    level, text = random.choice(messages)
    logging.log(level, f"{text} | Vorgang {i}")
````

## Best Practices Pakete / Module

Ein gutes Python-Paket ist so aufgebaut, dass

- jedes Modul eine klar abgegrenzte Aufgabe hat,
- Abhängigkeiten möglichst eindeutig verlaufen,
- die öffentliche Schnittstelle klein bleibt,
- Tests und Quellcode sauber getrennt sind,
- Imports unabhängig vom aktuellen Arbeitsverzeichnis funktionieren.

Dabei ist nicht das Ziel, möglichst viele Dateien zu erzeugen. Zu starke Aufteilung macht ein Projekt ebenso unübersichtlich wie eine einzige riesige Datei.

Faustregeln:
- Ein Modul sollte einen klaren Satz vervollständigen können: „Dieses Modul ist verantwortlich für …“
- Mit wenigen Modulen beginnen --> Module wachsen lassen --> Erst bei klar erkennbaren Verantwortlichkeiten aufteilen.
- Aussagekräftige Modulnamen verwenden
- Absolute Imports bevorzugen: innerhalb größerer Projekte sind absolute Imports oft am verständlichsten. Tiefe relative Imports vermeiden.

## pyproject.toml und moderne Python Projektstruktur

pyproject.toml ist die zentrale Konfigurationsdatei eines modernen Python-Projekts. Darin kann man unter anderem festlegen:

- wie das Projekt heißt und welche Version es hat,
- welche Python-Version benötigt wird,
- welche Abhängigkeiten installiert werden sollen,
- wie das Paket gebaut wird,
- wie Werkzeuge wie pytest, ruff, black oder mypy konfiguriert werden.

Die Datei liegt normalerweise im Wurzelverzeichnis des Projekts. Die drei wichtigsten Bereiche sind [build-system], [project] und werkzeugspezifische [tool.*]-Abschnitte

        taschenrechner/
        ├── pyproject.toml
        ├── src/
        │   └── taschenrechner/
        │       ├── __init__.py
        │       └── rechnen.py
        └── tests/
            └── test_rechnen.py


Das Projekt installieren: Im Verzeichnis, in dem pyproject.toml liegt:
    
        python -m pip install -e .

Der editierbare Modus bedeutet, dass Änderungen an den Dateien unter src/ sofort wirksam sind. Du musst das Projekt während der Entwicklung nicht nach jeder Änderung neu installieren.

Ein großer Vorteil: zentrale Werkzeugkonfiguration

Mehrere Werkzeuge können dieselbe Datei verwenden:

        [tool.pytest.ini_options]
        testpaths = ["tests"]

        [tool.black]
        line-length = 100

        [tool.ruff]
        line-length = 100

        [tool.mypy]
        python_version = "3.11"
        strict = true

Ohne pyproject.toml hättest du möglicherweise mehrere Konfigurationsdateien:

        pytest.ini
        setup.cfg
        mypy.ini
        .ruff.toml

Mit pyproject.toml kann vieles an einer zentralen Stelle stehen. Welche Einträge unter [tool.*] erlaubt sind, bestimmt allerdings jeweils das entsprechende Werkzeug.

## venv (nativ verfügbar)

| Zweck                                                  | Befehl                                                                                                | Erklärung                                                                                                              |
| ------------------------------------------------------ | ----------------------------------------------------------------------------------------------------- | ---------------------------------------------------------------------------------------------------------------------- |
| Virtuelle Umgebung erstellen                           | `python3 -m venv .venv`                                                                               | Erstellt im aktuellen Projektverzeichnis eine virtuelle Umgebung im Ordner `.venv`.                                    |
| Umgebung aktivieren, Linux/macOS                       | `source .venv/bin/activate`                                                                           | Aktiviert die Umgebung in der aktuellen Shell. Danach zeigen `python` und `pip` auf die virtuelle Umgebung.            |
| Umgebung aktivieren, Windows PowerShell                | `.venv\Scripts\Activate.ps1`                                                                          | Aktiviert die Umgebung unter Windows in PowerShell.                                                                    |
| Umgebung aktivieren, Windows CMD                       | `.venv\Scripts\activate.bat`                                                                          | Aktiviert die Umgebung in der klassischen Windows-Eingabeaufforderung.                                                 |
| Aktiven Python-Interpreter prüfen, Linux/macOS         | `which python`                                                                                        | Zeigt den Pfad des aktuell verwendeten Python-Interpreters an. Bei aktiver Umgebung sollte der Pfad in `.venv` liegen. |
| Aktiven Python-Interpreter prüfen, Windows             | `where python`                                                                                        | Windows-Variante von `which python`.                                                                                   |
| Python-Version anzeigen                                | `python --version`                                                                                    | Zeigt die Python-Version der aktuell verwendeten Umgebung.                                                             |
| `pip`-Version und Pfad prüfen                          | `python -m pip --version`                                                                             | Zeigt, welches `pip` verwendet wird und zu welchem Python es gehört.                                                   |
| Paket installieren                                     | `python -m pip install pytest`                                                                        | Installiert das Paket `pytest` in der aktiven virtuellen Umgebung.                                                     |
| Bestimmte Paketversion installieren                    | `python -m pip install pytest==8.3.5`                                                                 | Installiert exakt die angegebene Version.                                                                              |
| Paket aktualisieren                                    | `python -m pip install --upgrade pytest`                                                              | Aktualisiert ein bereits installiertes Paket.                                                                          |
| `pip` aktualisieren                                    | `python -m pip install --upgrade pip`                                                                 | Aktualisiert den Paketmanager `pip` innerhalb der Umgebung.                                                            |
| Paket deinstallieren                                   | `python -m pip uninstall pytest`                                                                      | Entfernt das Paket aus der virtuellen Umgebung.                                                                        |
| Installierte Pakete anzeigen                           | `python -m pip list`                                                                                  | Zeigt alle installierten Pakete und deren Versionen.                                                                   |
| Informationen zu einem Paket anzeigen                  | `python -m pip show pytest`                                                                           | Zeigt Details wie Version, Installationspfad und Abhängigkeiten.                                                       |
| Abhängigkeiten exportieren                             | `python -m pip freeze > requirements.txt`                                                             | Schreibt alle installierten Pakete mit ihren exakten Versionen in `requirements.txt`.                                  |
| Abhängigkeiten installieren                            | `python -m pip install -r requirements.txt`                                                           | Installiert alle Pakete, die in `requirements.txt` aufgeführt sind.                                                    |
| Umgebung verlassen                                     | `deactivate`                                                                                          | Deaktiviert die virtuelle Umgebung und wechselt zurück zum normalen System-Python.                                     |
| Umgebung löschen, Linux/macOS                          | `rm -rf .venv`                                                                                        | Löscht die komplette virtuelle Umgebung. Dafür gibt es keinen eigenen `venv`-Befehl.                                   |
| Umgebung löschen, Windows PowerShell                   | `Remove-Item -Recurse -Force .venv`                                                                   | Löscht die virtuelle Umgebung unter Windows PowerShell.                                                                |
| Umgebung neu erstellen                                 | `python3 -m venv .venv`                                                                               | Erstellt eine zuvor gelöschte Umgebung erneut.                                                                         |
| Umgebung mit vorhandenen Abhängigkeiten rekonstruieren | `python3 -m venv .venv`<br>`source .venv/bin/activate`<br>`python -m pip install -r requirements.txt` | Erstellt die Umgebung, aktiviert sie und installiert anschließend alle gespeicherten Abhängigkeiten.                   |

# Python-Dokumentation mit Sphinx erzeugen

## 1. Sphinx installieren

```bash
pip install sphinx
```

## 2. Sphinx-Grundgerüst erzeugen

Im Projektverzeichnis:

```bash
sphinx-quickstart docs
```

Dadurch wird u. a. folgende Struktur erzeugt:

```text
projekt/
├── docs/
│   ├── conf.py
│   ├── index.rst
│   └── Makefile
├── __init__.py
└── mein_modul.py
```

## 3. `docs/conf.py` konfigurieren

Python-Projekt in den Suchpfad aufnehmen:

```python
import os
import sys

sys.path.insert(0, os.path.abspath("../.."))
```

Sphinx-Erweiterungen aktivieren:

```python
extensions = [
    "sphinx.ext.autodoc",
    "sphinx.ext.napoleon",
]
```

- `autodoc`: erzeugt Dokumentation aus Python-Modulen, Klassen und Funktionen
- `napoleon`: unterstützt z. B. Google-Style- und NumPy-Style-Docstrings

## 4. Python-Code mit Docstrings versehen

Beispiel:

```python
def add(a: float, b: float) -> float:
    """
    Addiert zwei Zahlen.

    Args:
        a: Erste Zahl.
        b: Zweite Zahl.

    Returns:
        Summe der beiden Zahlen.
    """
    return a + b
```

## 5. API-Dokumentation automatisch erzeugen

Im Projektverzeichnis:

```bash
sphinx-apidoc -o docs/api .
```

Dadurch erzeugt Sphinx `.rst`-Dateien unter:

```text
docs/api/
```

z. B.:

```text
docs/api/modules.rst
docs/api/mein_paket.rst
```

## 6. API-Dokumentation in `docs/index.rst` einbinden

```rst
.. toctree::
   :maxdepth: 2
   :caption: Contents:

   api/modules
```

Ohne diesen Eintrag kann die Warnung

```text
document isn't included in any toctree
```

erscheinen.

## 7. HTML-Dokumentation erzeugen

```bash
cd docs
make html
```

Die erzeugte HTML-Dokumentation befindet sich anschließend unter:

```text
docs/_build/html/
```

## 8. Bei Bedarf komplett neu bauen

```bash
make clean
make html
```

## 9. Dokumentation unter Linux öffnen

```bash
xdg-open _build/html/index.html
```

## Gesamtablauf

```text
Python-Code + Docstrings
        ↓
sphinx-apidoc
        ↓
.rst-Dateien
        ↓
make html
        ↓
HTML-Dokumentation
```

Kurzfassung:

```bash
pip install sphinx
sphinx-quickstart docs
sphinx-apidoc -o docs/api .
cd docs
make html
xdg-open _build/html/index.html
```
