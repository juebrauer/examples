```python
from datetime import date
import meteostat as ms
import matplotlib.pyplot as plt

# Kempten: Breitengrad (Nord) und Längengrad (Ost).
breitengrad = 47.728569
laengengrad = 10.315784
ort = ms.Point(breitengrad, laengengrad)

# Nahegelegene Stationen suchen und deren Tagesdaten für Kempten kombinieren.
stationen = ms.stations.nearby(ort, limit=4)
zeitreihe = ms.daily(stationen,
                     date(2025, 12, 1),
                     date(2026, 12, 1))
daten = ms.interpolate(zeitreihe, ort).fetch()
daten.head()  # Die ersten fünf Tabellenzeilen ansehen.
```

```python
plt.figure(figsize=(12, 5))
plt.plot(daten["temp"])
plt.show()
```
