# Formatierte Strings

```python
import random
zufallszahl = random.randint(1,100_000)
print(f"Die Zufallszahl ist heute {zufallszahl:0>10}")
```

```python
# Dies ist meine 1. Version eines TAN-Generators in Python

import random

def tan_block_generator(rows, cols):
    for row in range(rows):
        for col in range(cols):
            zufallszahl = random.randint(1,1_000_000)
            print(f"{zufallszahl:0>7}", end=" ")
        print()

tan_block_generator(5,2)

print('-' * 30)

tan_block_generator(8,4)
```

# match-case in Python

```python
auswahl = int(input("Auswahl: "))

match auswahl:
 case 1 : N = 5
 case 2 : N = 10
 case 3 : N = 20
 case 4 : N = 100
 case _:
    print("Ungültige Auswahl!")

print(N)
```
