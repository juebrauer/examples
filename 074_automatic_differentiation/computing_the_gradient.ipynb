# Motivation: Grenzen des symbolischen Differenzierens

Was ist aber, wenn die Funktion als "Programm" gegeben ist, z.B. mit if-Anweisungen?

Dann versagt das symbolische Differenzieren.

````python
def f(x,y):
    return x*y * (x**2+y**2)

def g(x,y):
    return x*y * (x**3+y**3)

def h(x,y):
    """
    Funktionswert und Gradient hängt sehr stark davon ab,
    ob x < y ist oder nicht!
    """
    if x<y:
        return f(x,y)
    else:
        return g(x,y)
````

````python
def f(x,y):
    return x*y * (x**2+y**2)

def g(x,y):
    return x*y * (x**3+y**3)

def h(x,y):
    """
    Funktionswert und Gradient hängt sehr stark davon ab,
    ob x < y ist oder nicht!
    """
    if x<y:
        return f(x,y)
    else:
        loop_sum = 0
        for i in range(0,int(x)):
            loop_sum += g(x,y)
        return loop_sum
````

# Numerisches Differenzieren

````python
def f(x,y):
    return x*y * (x**2+y**2)

def df_dx(x,y, s=0.001):
    approx = ( f(x+s, y) - f(x,y) ) / s
    return approx

def df_dy(x,y, s=0.001):
    approx = ( f(x, y+s) - f(x,y) ) / s
    return approx
````

````python
df_dx(4,5)
````

````python
df_dy(4,5)
````

````python
def dh_dx(x,y, s=0.001):
    approx = ( h(x+s, y) - h(x,y) ) / s
    return approx

def dh_dy(x,y, s=0.001):
    approx = ( h(x, y+s) - h(x,y) ) / s
    return approx
````

# Forward-Mode AutoDiff in einem Graphen

Jeder Berechnungsknoten muss wissen
- welche Operation führt er aus? (+,-,*,^2,sin,...)
- von welchen Elternknoten/Vorgängern hängt er ab?

````python
# ==========================================================
# Forward-Mode AutoDiff mit einem Berechnungsgraphen
#
# f(x,y) = (x*y) * (x²+y²)
#
# Beispiel: x = 4, y = 5
# ==========================================================


class Node:
    def __init__(self, name, op=None, derivative_rule=None,
                 inputs=None, value=None, dot=None):

        self.name = name
        # Name des Knotens, z.B. "w1", "w2", ...

        self.op = op
        # Operation zur Berechnung des Knotenwertes
        # z.B. add, mult oder square
        # None bei Eingabeknoten

        self.derivative_rule = derivative_rule
        # Lokale Ableitungsregel der Operation
        # z.B. d_add, d_mult oder d_square
        # None bei Eingabeknoten

        self.inputs = inputs or []
        # Liste der direkten Vorgängerknoten
        # z.B. [w1, w2] bei einer Multiplikation

        self.value = value
        # Berechneter Wert des Knotens (w_i)
        # Bei Eingabeknoten wird dieser vorgegeben

        self.dot = dot
        # Ableitung des Knotens nach der gewählten
        # Eingangsvariablen (w_i mit Punkt darüber)
        # Bei Eingabeknoten: Saat-Variable (0 oder 1)


    def compute(self):
        # Eingabeknoten haben bereits Wert und Ableitung
        if self.op is None:
            return

        # Werte und Ableitungen der Vorgängerknoten holen
        values = [node.value for node in self.inputs]
        dots = [node.dot for node in self.inputs]

        # Funktionswert berechnen
        self.value = self.op(*values)

        # Ableitung mittels lokaler Ableitungsregel berechnen
        self.dot = self.derivative_rule(*values, *dots)


# ==========================================================
# Operationen und ihre Ableitungsregeln
# ==========================================================

# Addition
def add(a, b):
    return a + b

def d_add(a, b, da, db):
    return da + db


# Multiplikation (Produktregel)
def mult(a, b):
    return a * b

def d_mult(a, b, da, db):
    return da*b + a*db


# Quadrieren (Kettenregel)
def square(a):
    return a**2

def d_square(a, da):
    return 2*a*da


# ==========================================================
# Berechnungsgraph aufbauen
# ==========================================================

# Eingabeknoten: x=4, y=5
# Saat-Variablen für die Ableitung nach x
w1 = Node("w1", value=4, dot=1)
w2 = Node("w2", value=5, dot=0)

# Innere Knoten
w3 = Node("w3", op=square, derivative_rule=d_square, inputs=[w1])
w4 = Node("w4", op=square, derivative_rule=d_square, inputs=[w2])
w5 = Node("w5", op=mult,   derivative_rule=d_mult,   inputs=[w1, w2])
w6 = Node("w6", op=add,    derivative_rule=d_add,    inputs=[w3, w4])
w7 = Node("w7", op=mult,   derivative_rule=d_mult,   inputs=[w6, w5])

# Knoten in topologischer Reihenfolge
graph = [w1, w2, w3, w4, w5, w6, w7]


# ==========================================================
# Forward Mode: Ableitung nach x
# ==========================================================

print("Ableitung nach x:")

for node in graph:
    node.compute()
    print(f"{node.name}: Wert={node.value}, Ableitung={node.dot}")

print("df/dx =", w7.dot)


# ==========================================================
# Forward Mode: Ableitung nach y
# ==========================================================

# Nur die Saat-Variablen ändern
w1.dot = 0
w2.dot = 1

print("\nAbleitung nach y:")

for node in graph:
    node.compute()
    print(f"{node.name}: Wert={node.value}, Ableitung={node.dot}")

print("df/dy =", w7.dot)
````

# Reverse-Mode AutoDiff in einem Graphen

````python
# ==========================================================
# Reverse-Mode AutoDiff mit einem Berechnungsgraphen
#
# f(x,y) = (x*y) * (x²+y²)
#
# Beispiel: x = 4, y = 5
# ==========================================================


class Node:
    def __init__(self, name, op=None, derivative_rule=None,
                 inputs=None, value=None):

        self.name = name
        # Name des Knotens, z.B. "w1", "w2", ...

        self.op = op
        # Operation zur Berechnung des Knotenwertes
        # z.B. add, mult oder square
        # None bei Eingabeknoten

        self.derivative_rule = derivative_rule
        # Lokale Ableitungsregel der Operation
        # Liefert die partiellen Ableitungen nach
        # den direkten Eingängen des Knotens

        self.inputs = inputs or []
        # Liste der direkten Vorgängerknoten
        # z.B. [w1, w2] bei einer Multiplikation

        self.value = value
        # Berechneter Wert des Knotens (w_i)
        # Bei Eingabeknoten wird dieser vorgegeben

        self.bar = 0
        # Adjungierte des Knotens:
        # bar(w_i) = dw7/dw_i
        # Sammelt die Ableitungsbeiträge aller Nachfolger


    def compute(self):
        # Vorwärtsschritt: Funktionswerte berechnen

        if self.op is None:
            return

        values = [node.value for node in self.inputs]

        self.value = self.op(*values)


    def backward(self):
        # Rückwärtsschritt: Ableitungen propagieren

        if self.op is None:
            return

        values = [node.value for node in self.inputs]

        # Lokale partielle Ableitungen berechnen
        derivatives = self.derivative_rule(*values)

        # Ableitungsbeiträge an Vorgänger verteilen
        for node, derivative in zip(self.inputs, derivatives):
            node.bar += self.bar * derivative


# ==========================================================
# Operationen und lokale Ableitungsregeln
# ==========================================================

# Addition
def add(a, b):
    return a + b

def d_add(a, b):
    return (1, 1)


# Multiplikation (Produktregel)
def mult(a, b):
    return a * b

def d_mult(a, b):
    return (b, a)


# Quadrieren
def square(a):
    return a**2

def d_square(a):
    return (2*a,)


# ==========================================================
# Berechnungsgraph aufbauen
# ==========================================================

# Eingabeknoten: x=4, y=5
w1 = Node("w1", value=4)
w2 = Node("w2", value=5)

# Innere Knoten
w3 = Node("w3", op=square, derivative_rule=d_square, inputs=[w1])
w4 = Node("w4", op=square, derivative_rule=d_square, inputs=[w2])
w5 = Node("w5", op=mult, derivative_rule=d_mult, inputs=[w1, w2])
w6 = Node("w6", op=add, derivative_rule=d_add, inputs=[w3, w4])
w7 = Node("w7", op=mult, derivative_rule=d_mult, inputs=[w6, w5])

# Knoten in topologischer Reihenfolge
graph = [w1, w2, w3, w4, w5, w6, w7]


# ==========================================================
# 1. Vorwärtsschritt: Werte berechnen
# ==========================================================

print("Vorwärtsschritt:")

for node in graph:
    node.compute()
    print(f"{node.name}: Wert={node.value}")


# ==========================================================
# 2. Rückwärtsschritt: Ableitungen berechnen
# ==========================================================

# Alle Adjungierten sind zunächst 0
for node in graph:
    node.bar = 0

# Saat-Variable am Output
w7.bar = 1

# Graph in umgekehrter Reihenfolge durchlaufen
for node in reversed(graph):
    node.backward()

print("\nRückwärtsschritt:")

for node in graph:
    print(f"{node.name}: bar={node.bar}")

print()
print("df/dx =", w1.bar)
print("df/dy =", w2.bar)
````

### Reverse-Mode AutoDiff: Wie funktioniert `backward()`?

Die zentrale Idee des Reverse Mode ist, dass jeder Knoten seine **Adjungierte** an seine Vorgängerknoten weitergibt. Dabei wird die Kettenregel angewendet.

Die Adjungierte eines Knotens $w_i$ ist definiert als:

$$
\bar{w}_i = \frac{\partial w_7}{\partial w_i}
$$

Betrachten wir beispielsweise den Multiplikationsknoten:

$$
w_5 = w_1 \cdot w_2
$$

Die lokale Ableitungsregel lautet:

```python
def d_mult(a, b):
    return (b, a)
```

Denn die beiden partiellen Ableitungen sind:

$$
\frac{\partial w_5}{\partial w_1} = w_2
\qquad
\frac{\partial w_5}{\partial w_2} = w_1
$$

Die `backward()`-Methode führt damit sinngemäß folgende Berechnungen aus:

```python
w1.bar += w5.bar * w2.value
w2.bar += w5.bar * w1.value
```

Für unser Beispiel mit $w_1=4$, $w_2=5$ und $\bar{w}_5=41$ ergeben sich die Beiträge:

$$
\begin{aligned}
\bar{w}_1 &\mathrel{+}= 41 \cdot 5 = 205\\
\bar{w}_2 &\mathrel{+}= 41 \cdot 4 = 164
\end{aligned}
$$

**Warum verwenden wir `+=` statt `=`?**

Die Knoten $w_1$ und $w_2$ beeinflussen den Output über mehrere Pfade. Deshalb müssen die Ableitungsbeiträge aller Pfade **aufsummiert** werden.

Über die Knoten $w_3=w_1^2$ und $w_4=w_2^2$ kommen noch weitere Beiträge hinzu:

$$
\begin{aligned}
\bar{w}_1 &= 205 + 20\cdot 8 = \boxed{365}\\
\bar{w}_2 &= 164 + 20\cdot 10 = \boxed{364}
\end{aligned}
$$

Damit erhalten wir den gesamten Gradienten:

$$
\nabla f(4,5) =
\begin{pmatrix}
365\\
364
\end{pmatrix}
$$

**Merksatz:** Im Reverse Mode verteilt jeder Knoten seine Adjungierte mithilfe der lokalen Ableitungsregeln an seine Vorgänger. Beiträge aus unterschiedlichen Pfaden werden dabei addiert.

Der entscheidende Code dafür lautet:

```python
for node, derivative in zip(self.inputs, derivatives):
    node.bar += self.bar * derivative
```

# Reverse-Mode AutoDiff mit PyTorch

Wir berechnen wieder $f(x,y)=(x\cdot y)\cdot(x^2+y^2)$ an der Stelle $(4,5)$.
PyTorch baut den Berechnungsgraphen **während des Vorwärtspasses automatisch auf**.
Die lokalen Ableitungsregeln für `+`, `*` und `**2` sind bereits implementiert.
Unsere eigene `Node`-Klasse brauchen wir deshalb nicht mehr.

## 1. Eingaben und Berechnungsgraph aufbauen

`requires_grad=True` markiert die Eingaben, nach denen wir ableiten möchten.
Dafür verwenden wir Fließkommazahlen (`4.0`, `5.0`), keine Integer-Tensoren.
Die Namen $w_1$ bis $w_7$ entsprechen genau unserem bisherigen Beispiel:

```text
w1 = x ──→ w3 = x² ──┐
                      ├──→ w6 = x² + y² ──┐
w2 = y ──→ w4 = y² ──┘                     │
                                            ├──→ w7 = f(x,y)
w1, w2 ────────────────→ w5 = x · y ────────┘
```

Falls PyTorch im Notebook-Kernel noch fehlt: einmal `%pip install torch` ausführen.

````python
import torch

# Blattknoten: direkt erzeugte Eingaben, deren Gradienten wir benötigen.
x = torch.tensor(4.0, dtype=torch.float64, requires_grad=True)
y = torch.tensor(5.0, dtype=torch.float64, requires_grad=True)

# Forward Pass: Werte berechnen UND den Autograd-Graphen aufbauen.
w1 = x
w2 = y
w3 = w1**2
w4 = w2**2
w5 = w1 * w2
w6 = w3 + w4
w7 = w6 * w5

nodes_torch = {"w1": w1, "w2": w2, "w3": w3, "w4": w4,
               "w5": w5, "w6": w6, "w7": w7}

print(f"f(4, 5) = {w7.item():g}")
print(f"Gradient vor backward(): x.grad = {x.grad}, y.grad = {y.grad}")
````

## 2. Was hat PyTorch gespeichert?

`grad_fn` verweist bei einem berechneten Tensor auf seine Rückwärtsoperation,
beispielsweise `MulBackward0` für eine Multiplikation. Über `next_functions`
sind diese Operationen mit ihren Vorgängern verbunden. Die genauen Klassennamen
können je nach PyTorch-Version variieren.

Die Eingaben `x` und `y` sind **Blattknoten** (`is_leaf=True`): Sie wurden direkt
erzeugt und haben deshalb keine erzeugende Operation (`grad_fn=None`).

````python
print(f"{'Knoten':<8} {'Wert':>8} {'Blatt?':>8}  Rückwärtsoperation")
print("-" * 58)
for name, tensor in nodes_torch.items():
    operation = type(tensor.grad_fn).__name__ if tensor.grad_fn else "— (Eingabe)"
    print(f"{name:<8} {tensor.item():>8g} {str(tensor.is_leaf):>8}  {operation}")

print("\nDirekte Vorgänger der Rückwärtsoperation von w7:")
for predecessor, _ in w7.grad_fn.next_functions:
    print("  →", type(predecessor).__name__)
````

## 3. Rückwärtspass starten und Gradienten anzeigen

Für den skalaren Output startet `w7.backward()` mit der Saat
$\bar w_7=\partial w_7/\partial w_7=1$ und summiert die Beiträge entlang aller Pfade.
Danach stehen die gesuchten Ableitungen in `x.grad` und `y.grad`.

Standardmäßig speichert PyTorch `.grad` nur für die Blattknoten mit
`requires_grad=True`. Für unsere Lehrdemo rufen wir **vor** `backward()` zusätzlich
`retain_grad()` auf den Zwischenwerten auf. So können wir auch deren Adjungierte
anzeigen. Für die Berechnung des Eingabegradienten ist das nicht erforderlich.

````python
# Auch die Adjungierten der Zwischenknoten für die Anzeige aufbewahren.
for tensor in nodes_torch.values():
    if not tensor.is_leaf:
        tensor.retain_grad()

# Genau ein Rückwärtspass liefert beide partiellen Ableitungen.
w7.backward()

gradient_torch = torch.stack((x.grad, y.grad))
print(f"∂f/∂x = {x.grad.item():g}")
print(f"∂f/∂y = {y.grad.item():g}")
print("∇f(4, 5) =", gradient_torch.tolist())

print(f"\n{'Knoten':<8} {'Wert':>8} {'Adjungierte ∂f/∂wi':>22}")
print("-" * 42)
for name, tensor in nodes_torch.items():
    print(f"{name:<8} {tensor.item():>8g} {tensor.grad.item():>22g}")

# Vergleich mit der analytischen Ableitung.
x_value, y_value = x.item(), y.item()
expected_gradient = torch.tensor([
    3 * x_value**2 * y_value + y_value**3,
    x_value**3 + 3 * x_value * y_value**2,
], dtype=torch.float64)
torch.testing.assert_close(gradient_torch, expected_gradient)
print("\n✓ Gradient stimmt mit der analytischen Ableitung überein.")
````

Damit erhalten wir denselben Funktionswert und Gradienten wie zuvor:

$$f(4,5)=820,\qquad \nabla f(4,5)=\begin{pmatrix}365\\364\end{pmatrix}.$$

**Beim erneuten Ausführen:** Starte wieder bei der ersten Codezelle dieses Abschnitts.
So entstehen neue Eingabetensoren und ein neuer Graph. `backward()` akkumuliert
Gradienten in `.grad`; vorhandene Gradienten werden nicht automatisch zurückgesetzt.
Außerdem werden standardmäßig die für den Rückwärtspass gespeicherten Zwischenwerte
freigegeben. Die `backward()`-Zelle allein erneut auszuführen ist deshalb hier nicht vorgesehen.

`retain_grad()` speichert die Adjungierten der Zwischenknoten; es ist nicht dasselbe
wie `retain_graph=True`, das die für einen weiteren Rückwärtspass benötigten Daten erhält.

und [Blattknoten und retain_grad](https://docs.pytorch.org/tutorials/beginner/understanding_leaf_vs_nonleaf_tutorial.html).
