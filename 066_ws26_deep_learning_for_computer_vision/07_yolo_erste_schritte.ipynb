# Inferenz auf einem Bild durchführen

```python
folder = "/home/juebrauer/link_to_vcd/08_datasets/67_parkplatz_bilder/"
img_name = "parkplatz3.jpg"
abs_fname = f"{folder}{img_name}"

from ultralytics import YOLO

# Vortrainiertes Modell laden
model = YOLO("yolo26x.pt")

# Autos im Parkplatzfoto erkennen
results = model.predict(
    source=abs_fname,
    classes=[2],  # COCO-Klasse 2 = car
    conf=0.01
)

# Anzahl erkannter Autos
anzahl = len(results[0].boxes)

print(f"Anzahl Autos: {anzahl}")

# Bild mit Bounding Boxes speichern
results[0].save(filename="ergebnis.jpg")
```
